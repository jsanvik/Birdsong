When given the dataset, I saw lattitute and longitude, and my first instinct was to plot them to visualize where the birds all are.  
Later, when given the manuscript to read, it was cool to see that the neighbourhoods I plotted here align closely with the map presented in the manuscript.

```r
# Set data types and select input columns
data <- data %>% 
    mutate(across(contains("UST"), ~ as.factor(.x)), 
        does_it_match = as.factor(does_it_match),
        number_of_matches = as.integer(number_of_matches))
```

```r
# Now plotting the above graphs using for loops to keep code concise
# Flight paths, colour by neighbourhood (shape by individual, not shown in legend)
fig = ggplot()
for (i in 1:6) {
    fig <- fig + 
        geom_path(data = data, aes_string(y = sprintf("M%s_Lat", i), x = sprintf("M%s_Lon", i), color = "neighbourhood"), shape = i) +
        geom_point(data = data, aes_string(y = sprintf("M%s_Lat", i), x = sprintf("M%s_Lon", i), color = "neighbourhood"), shape = i)
}
# sprintf
```
