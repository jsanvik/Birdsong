```r
# Flight paths, colour by neighbourhood (shape by individual, not shown in legend)
ggplot() +
    geom_path(data = data, aes(y = M1_Lat, x = M1_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M1_Lat, x = M1_Lon, color = neighbourhood), shape = 1) +
    geom_path(data = data, aes(y = M2_Lat, x = M2_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M2_Lat, x = M2_Lon, color = neighbourhood), shape = 2) +
    geom_path(data = data, aes(y = M3_Lat, x = M3_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M3_Lat, x = M3_Lon, color = neighbourhood), shape = 3) +
    geom_path(data = data, aes(y = M4_Lat, x = M4_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M4_Lat, x = M4_Lon, color = neighbourhood), shape = 4) +
    geom_path(data = data, aes(y = M5_Lat, x = M5_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M5_Lat, x = M5_Lon, color = neighbourhood), shape = 5) +
    geom_path(data = data, aes(y = M6_Lat, x = M6_Lon, color = neighbourhood)) +
    geom_point(data = data, aes(y = M6_Lat, x = M6_Lon, color = neighbourhood), shape = 6) +
    ylab("Latitude") +
    xlab("Longitude")
```

```r
# Now plotting the above graphs using for loops to keep code concise
# Flight paths, colour by neighbourhood (shape by individual, not shown in legend)
fig = ggplot()
for (i in 1:6) {
    fig <- fig + 
        geom_path(data = data, aes_string(y = sprintf("M%s_Lat", i), x = sprintf("M%s_Lon", i), color = "neighbourhood"), shape = i) +
        geom_point(data = data, aes_string(y = sprintf("M%s_Lat", i), x = sprintf("M%s_Lon", i), color = "neighbourhood"), shape = i)
}
fig
# sprintf
```
