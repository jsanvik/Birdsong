```r
folder_path <- "C:\\Users\\johna\\OneDrive\\Documents\\Birdsong"

folder_paths <- list.files(path = folder_path, 
                          full.names = TRUE)

# Check if any files were found

for (path in folder_paths) {
    if (!dir.exists(path)) {
        cat(path, " not a folder. \n")
    } else if (length(path) == 0) {
    cat("No Jupyter Notebooks found in the specified directory.\n")
    } else {
    

    new_path <- list.files(path = path, 
                            pattern = "\\.ipynb$", 
                            full.names = TRUE)
    cat(path, "\n", "Found", length(new_path), "notebooks. Starting conversion...\n")

    for (file in new_path) {
        # Generate the output filename (replaces .ipynb with .Rmd)
        output_file <- sub("\\.ipynb$", ".Rmd", file)
        print(file)

        tryCatch({
            convert_ipynb(file, output = output_file)
            cat("Successfully converted:", basename(file), "\n")
        }, error = function(e) {
            cat("Error converting", basename(file), ":", conditionMessage(e), "\n")
        })
  }
    }
  
}

cat("\nConversion process complete.\n")
```
