```r
library(tidyverse)
data <- read_csv("../data/minizord.csv")
```

```r
TIME_INTERVAL <- "30 seconds"

time_data <- data %>%
    mutate(interval = floor_date(DateTime, unit=TIME_INTERVAL)) %>%
    group_by(interval) %>%
    summarise(
        # Distance
        across(where((is.numeric)), ~mean(.x, na.rm=FALSE))
        # Match Values
    )

time_data
```
